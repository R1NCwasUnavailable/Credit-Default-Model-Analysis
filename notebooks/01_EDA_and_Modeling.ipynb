# Credit Card Default Analysis - EDA & Model Demo

This notebook demonstrates the exploratory data analysis and basic model training workflow used in the project.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_loader import load_data
from src.preprocessing import clean_data

# Load Data
df = load_data()
df_clean = clean_data(df)
```

## 1. Exploratory Data Analysis

```python
# Target Distribution
plt.figure(figsize=(6,4))
sns.countplot(x='default payment next month', data=df_clean)
plt.title('Target Distribution (Default vs Non-Default)')
plt.show()
```

```python
# Correlation Matrix
plt.figure(figsize=(15,10))
sns.heatmap(df_clean.corr(), annot=False, cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.show()
```

## 2. Model Training Demo
Here we train a simple Random Forest model as a quick demonstration.

```python
from src.preprocessing import preprocess_data
from sklearn.ensemble import RandomForestClassifier
from src.evaluation import evaluate_model

X_train, X_test, y_train, y_test = preprocess_data(df)

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

metrics = evaluate_model(rf, X_test, y_test, "Random Forest Demo")
print(metrics)
```
